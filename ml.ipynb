## Data quality and variable distributions

These checks report missing values, duplicate rows/IDs, category frequencies, and blood-pressure values that may need review. Outlier flags are prompts for investigation, not automatic exclusion rules; the source data is left unchanged.

```python
from pathlib import Path
from IPython.display import display

plot_dir = Path("plot")
plot_dir.mkdir(parents=True, exist_ok=True)

numerical_variables = ["age_years", "height", "weight", "ap_hi", "ap_lo"]
categorical_variables = ["gender", "cholesterol", "gluc", "smoke", "alco", "active"]

print(f"Rows: {len(data):,}; columns: {data.shape[1]}")
print(f"Duplicate rows: {data.duplicated().sum():,}")
print(f"Duplicate IDs: {data['id'].duplicated().sum():,}")
display(pd.DataFrame({
    "dtype": data.dtypes.astype(str),
    "missing_values": data.isna().sum(),
}))

bp_checks = pd.DataFrame({
    "check": [
        "Rows with non-positive systolic or diastolic pressure",
        "Rows where diastolic pressure is at least systolic pressure",
    ],
    "rows_flagged": [
        ((data["ap_hi"] <= 0) | (data["ap_lo"] <= 0)).sum(),
        (data["ap_lo"] >= data["ap_hi"]).sum(),
    ],
})
display(bp_checks)

bp_outlier_rows = []
for variable in ["ap_hi", "ap_lo"]:
    q1, q3 = data[variable].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr
    outside_fences = (data[variable] < lower_fence) | (data[variable] > upper_fence)
    bp_outlier_rows.append({
        "variable": variable,
        "lower_IQR_fence": lower_fence,
        "upper_IQR_fence": upper_fence,
        "values_outside_fences": outside_fences.sum(),
    })
display(pd.DataFrame(bp_outlier_rows).round(2))

print("Numerical-variable summary:")
display(data[numerical_variables].describe().T.round(2))

category_counts = pd.concat(
    [
        data[variable].value_counts(dropna=False).rename_axis("code").reset_index(name="count").assign(variable=variable)
        for variable in categorical_variables + ["cardio"]
    ],
    ignore_index=True,
)[["variable", "code", "count"]]
print("Categorical feature and target counts:")
display(category_counts)

target_summary = data["cardio"].value_counts().sort_index().rename_axis("cardio").to_frame("count")
target_summary["percent"] = 100 * target_summary["count"] / len(data)
display(target_summary.round(2))
```

## Plotting `cardio` against other variables

The plots below compare numerical variables between the two `cardio` groups using horizontal boxplots, show outcome proportions for categorical variables, and visualize Spearman correlations for continuous measurements and the binary target. Each plot is saved as a PNG in `plot/` relative to the notebook's working directory. The record identifier (`id`) and integer-coded categories are excluded from the correlation calculation.

```python
from pathlib import Path

plot_dir = Path("plot")
plot_dir.mkdir(parents=True, exist_ok=True)

numerical_variables = ["age_years", "height", "weight", "ap_hi", "ap_lo"]
categorical_variables = ["gender", "cholesterol", "gluc", "smoke", "alco", "active"]
numerical_labels = {
    "age_years": "Age (years)",
    "height": "Height (cm)",
    "weight": "Weight (kg)",
    "ap_hi": "Systolic blood pressure (mmHg)",
    "ap_lo": "Diastolic blood pressure (mmHg)",
}

for variable in numerical_variables:
    plot_data = pd.DataFrame({
        "cardio": data["cardio"],
        "value": data[variable],
    })
    fig, ax = plt.subplots(figsize=(8, 5))
    sb.boxplot(
        data=plot_data,
        x="value",
        y="cardio",
        order=[0, 1],
        orient="h",
        showfliers=False,
        ax=ax,
    )
    ax.set_title(f"{numerical_labels[variable]} by cardiovascular disease status")
    ax.set_xlabel(numerical_labels[variable])
    ax.set_ylabel("Cardiovascular disease (0 = no, 1 = yes)")
    ax.set_yticks([0, 1], labels=["No", "Yes"])
    fig.tight_layout()
    fig.savefig(plot_dir / f"cardio_{variable}.png", dpi=150)
    plt.close(fig)

for variable in categorical_variables:
    proportions = pd.crosstab(
        data[variable], data["cardio"], normalize="index"
    ).reindex(columns=[0, 1], fill_value=0)
    proportions = proportions.rename(columns={
        0: "No cardiovascular disease",
        1: "Cardiovascular disease",
    })
    fig, ax = plt.subplots(figsize=(7, 5))
    proportions.plot(kind="bar", stacked=True, ax=ax)
    ax.set_title(f"Cardiovascular disease proportion by {variable}")
    ax.set_xlabel(variable.replace("_", " ").title())
    ax.set_ylabel("Proportion of records")
    ax.set_ylim(0, 1)
    ax.legend(title="Outcome")
    fig.tight_layout()
    fig.savefig(plot_dir / f"cardio_vs_{variable}.png", dpi=150)
    plt.close(fig)

correlation_variables = numerical_variables + ["cardio"]
correlation_matrix = data[correlation_variables].corr(method="spearman")
fig, ax = plt.subplots(figsize=(9, 7))
sb.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5,
    ax=ax,
)
ax.set_title("Spearman correlation: continuous variables and cardio")
fig.tight_layout()
fig.savefig(plot_dir / "correlation_heatmap.png", dpi=150)
plt.close(fig)

print(f"Saved {len(numerical_variables) + len(categorical_variables) + 1} plots to {plot_dir.resolve()}")
```

## Cardio prevalence by age and gender code

These are descriptive outcome prevalence comparisons, not model sensitivity/recall. The dataset documentation does not provide a verified mapping from gender codes to labels, so the groups remain identified by code.

```python
age_group_labels = ["20-29", "30-39", "40-49", "50-59", "60-69"]
subgroup_data = data.assign(
    age_group=pd.cut(
        data["age_years"],
        bins=[20, 30, 40, 50, 60, 70],
        right=False,
        labels=age_group_labels,
    )
)
age_summary = (
    subgroup_data.groupby("age_group", observed=True)
    .agg(n=("cardio", "size"), prevalence=("cardio", "mean"))
    .reset_index()
)
gender_summary = (
    data.groupby("gender")
    .agg(n=("cardio", "size"), prevalence=("cardio", "mean"))
    .reset_index()
)
gender_summary["group"] = gender_summary["gender"].map(lambda code: f"Gender code {code}")
display(age_summary.assign(prevalence=age_summary["prevalence"].map(lambda value: f"{value:.1%}")))
display(gender_summary[["group", "n", "prevalence"]].assign(
    prevalence=gender_summary["prevalence"].map(lambda value: f"{value:.1%}")
))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, summary, group_column, title in [
    (axes[0], age_summary, "age_group", "By age group"),
    (axes[1], gender_summary, "group", "By gender code"),
]:
    bars = ax.bar(summary[group_column].astype(str), summary["prevalence"])
    labels = [
        f"{rate:.1%} (n={count:,})"
        for rate, count in zip(summary["prevalence"], summary["n"])
    ]
    ax.bar_label(bars, labels=labels, padding=3, fontsize=8)
    ax.set_title(title)
    ax.set_xlabel("Group")
    ax.set_ylabel("Cardio prevalence")
    ax.set_ylim(0, 0.85)
    ax.tick_params(axis="x", rotation=20)
fig.suptitle("Observed cardiovascular disease prevalence by subgroup")
fig.tight_layout()
fig.savefig(plot_dir / "cardio_prevalence_by_age_gender.png", dpi=150)
plt.close(fig)
```

## EDA notes and interpretation

- The CSV has 70,000 records, no missing values, and no duplicate rows or IDs. The target is nearly balanced (35,021 records with `cardio = 0`; 34,979 with `cardio = 1`).
- Blood-pressure data needs review before it is used for a clinical-rule comparison: 29 records have a non-positive systolic or diastolic value, 1,236 have diastolic pressure greater than or equal to systolic pressure, and the raw maxima are 16,020 and 11,000 mmHg. IQR fences in the checks above flag distributional outliers, not confirmed errors. Do not silently remove or correct records; agree on a documented rule with the team and report how it affects the analysis.
- In this sample, cardio prevalence increases across the age groups with adequate sample sizes: 23.9% at ages 30-39, 38.0% at 40-49, 51.9% at 50-59, and 67.0% at 60-69. The 20-29 group contains only 3 records, so its estimate is not reliable. These are associations in this dataset, not causal effects.
- Prevalence is 49.7% for gender code 1 and 50.5% for code 2. Because the supplied description does not verify the code-to-sex mapping, do not label these as male/female until the mapping is confirmed. These prevalence rates do not answer whether model sensitivity differs by group; that requires held-out model predictions and subgroup recall in the model evaluation.
- The unadjusted categorical plots show higher cardio prevalence at higher recorded cholesterol (44.0%, 60.2%, and 76.5% for codes 1-3) and glucose (48.1%, 59.3%, and 62.2% for codes 1-3). Interpret these descriptively, not as causal or adjusted effects.
- The heatmap uses Spearman correlations for continuous measurements and the binary target, leaving out `id` and integer-coded categories. Blood-pressure anomalies can still influence its results, so interpret those correlations cautiously. Correlation does not establish causation.
- Agree with the ML team on the blood-pressure rule and its clinical source/threshold before comparing it with a model. Keep the final comparison and age-/sex-group sensitivity analysis on the agreed evaluation data; do not choose a rule from test-set results.

## Data cleaning and creating new health variables

1. filtering out records with biologically impossible blood pressure, height, and weight values.
2. create new clinical features: Body Mass Index (BMI), pulse pressure, and 10-year age groups.

```python
if 'age_years' not in data.columns:
    data['age_years'] = (data['age'] / 365.25).astype(int)

data_clean = data[
    (data['ap_hi'] >= 70) & (data['ap_hi'] <= 240) &
    (data['ap_lo'] >= 40) & (data['ap_lo'] <= 150) &
    (data['ap_hi'] > data['ap_lo']) &
    (data['height'] >= 120) & (data['height'] <= 220) &
    (data['weight'] >= 30) & (data['weight'] <= 200)
].copy()

data_clean['bmi'] = data_clean['weight'] / ((data_clean['height'] / 100) ** 2)
data_clean['pulse_pressure'] = data_clean['ap_hi'] - data_clean['ap_lo']

data_clean['age_group'] = pd.cut(
    data_clean['age_years'],
    bins=[30, 40, 50, 60, 100],
    labels=['30-39', '40-49', '50-59', '60-69'],
    right=False
)

print("Original dataset dims :", data.shape)
print("Cleaned dataset dims  :", data_clean.shape)
print("Removed outlier rows  :", len(data) - len(data_clean))
```

## Train and Test Split

Extract non-invasive features as predictor DataFrame X and target DataFrame y, splitting the dataset into 80% training and 20% testing partitions.

```python
from sklearn.model_selection import train_test_split

feature_cols = [
    'age_years', 'gender', 'height', 'weight', 
    'ap_hi', 'ap_lo', 'smoke', 'alco', 'active', 
    'bmi', 'pulse_pressure'
]

X = pd.DataFrame(data_clean[feature_cols])
y = pd.DataFrame(data_clean['cardio'])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Train set dims :", X_train.shape, y_train.shape)
print("Test set dims  :", X_test.shape, y_test.shape)
```

## Decision Tree Classifier

Training a Decision Tree Classifier **max depth = 5** on the training set and evaluate its classification accuracy, confusion matrix, and tree structure.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree

dectree = DecisionTreeClassifier(max_depth=5, random_state=42)
dectree.fit(X_train, y_train)

y_train_pred = dectree.predict(X_train)
y_test_pred = dectree.predict(X_test)

cm_train = confusion_matrix(y_train, y_train_pred)
cm_test = confusion_matrix(y_test, y_test_pred)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

disp_train = ConfusionMatrixDisplay(confusion_matrix=cm_train, display_labels=["No Cardio", "Cardio"])
disp_train.plot(ax=axes[0], cmap="Blues")
axes[0].set_title("Train Confusion Matrix")
axes[0].set_xlabel("Predicted Label")
axes[0].set_ylabel("Actual Label")

disp_test = ConfusionMatrixDisplay(confusion_matrix=cm_test, display_labels=["No Cardio", "Cardio"])
disp_test.plot(ax=axes[1], cmap="Oranges")
axes[1].set_title("Test Confusion Matrix")
axes[1].set_xlabel("Predicted Label")
axes[1].set_ylabel("Actual Label")

plt.tight_layout()
plt.show()

plt.figure(figsize=(18, 7))
plot_tree(
    dectree,
    feature_names=feature_cols,
    class_names=["No Cardio", "Cardio"],
    filled=True,
    max_depth=3,
    fontsize=9,
)
plt.title("Decision Tree Decision Structure (Top 3 Levels)")
plt.tight_layout()
plt.show()
```

## Model Benchmarking

We benchmark the Decision Tree against a Random Forest Classifier and a standard clinical baseline rule (Systolic>= 140 and Diastolic >= 90).

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred_doctor = ((X_test['ap_hi'] >= 140) | (X_test['ap_lo'] >= 90)).astype(int)

rf_model = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42)
rf_model.fit(X_train, y_train.values.ravel())
y_test_pred_rf = rf_model.predict(X_test)

benchmark_df = pd.DataFrame({
    'Method': ['Clinical Baseline Rule', 'Decision Tree', 'Random Forest'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_doctor),
        accuracy_score(y_test, y_test_pred),
        accuracy_score(y_test, y_test_pred_rf)
    ],
    'Precision': [
        precision_score(y_test, y_pred_doctor),
        precision_score(y_test, y_test_pred),
        precision_score(y_test, y_test_pred_rf)
    ],
    'Recall': [
        recall_score(y_test, y_pred_doctor),
        recall_score(y_test, y_test_pred),
        recall_score(y_test, y_test_pred_rf)
    ],
    'F1-Score': [
        f1_score(y_test, y_pred_doctor),
        f1_score(y_test, y_test_pred),
        f1_score(y_test, y_test_pred_rf)
    ]
}).round(4)

benchmark_df
```

## Subgroup Recall Analysis Across Age Groups

Comparing sensitivity (Recall) across 10-year age groups to test whether detection performance remains balanced across patient age cohorts.

```python
test_df = data_clean.loc[X_test.index].copy()
test_df['y_true'] = y_test.values
test_df['pred_doc'] = y_pred_doctor
test_df['pred_dt'] = y_test_pred
test_df['pred_rf'] = y_test_pred_rf

age_groups = ['30-39', '40-49', '50-59', '60-69']
age_results = []

for grp in age_groups:
    sub = test_df[test_df['age_group'] == grp]
    if len(sub) > 0:
        age_results.append({'Subgroup': grp, 'Method': 'Clinical Rule', 'Recall': recall_score(sub['y_true'], sub['pred_doc'])})
        age_results.append({'Subgroup': grp, 'Method': 'Decision Tree', 'Recall': recall_score(sub['y_true'], sub['pred_dt'])})
        age_results.append({'Subgroup': grp, 'Method': 'Random Forest', 'Recall': recall_score(sub['y_true'], sub['pred_rf'])})

df_age_recall = pd.DataFrame(age_results)

gender_map = {1: 'Women', 2: 'Men'}
gender_results = []

for g_code, g_name in gender_map.items():
    sub = test_df[test_df['gender'] == g_code]
    if len(sub) > 0:
        gender_results.append({'Sex Group': g_name, 'Method': 'Clinical Rule', 'Recall': recall_score(sub['y_true'], sub['pred_doc'])})
        gender_results.append({'Sex Group': g_name, 'Method': 'Decision Tree', 'Recall': recall_score(sub['y_true'], sub['pred_dt'])})
        gender_results.append({'Sex Group': g_name, 'Method': 'Random Forest', 'Recall': recall_score(sub['y_true'], sub['pred_rf'])})

df_gender_recall = pd.DataFrame(gender_results)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sb.barplot(data=df_age_recall, x='Subgroup', y='Recall', hue='Method', palette='Set2', ax=axes[0])
axes[0].set_title('Subgroup Recall Across Age Groups')
axes[0].set_ylim(0, 1.0)
axes[0].axhline(0.70, color='r', linestyle='--')

sb.barplot(data=df_gender_recall, x='Sex Group', y='Recall', hue='Method', palette='Set2', ax=axes[1])
axes[1].set_title('Subgroup Recall Across Sex Groups')
axes[1].set_ylim(0, 1.0)
axes[1].axhline(0.70, color='r', linestyle='--')

plt.tight_layout()
plt.show()
```

## Notes and Interpretation
Both non-invasive machine learning models outperform the standard clinical baseline rule (ap_hi >= 140 or ap_lo >=90) across overall evaluation metrics. The Random Forest classifier achieves the highest overall accuracy (72.16%) and F1-score (0.7007), while the Decision Tree yields the highest overall screening sensitivity (67.56% recall vs. 54.80% for the clinical rule). Combining multi-variable non-invasive features provides a 11 to 13 percentage point improvement in recall over blood-pressure thresholding alone, significantly reducing false negatives in frontline screening.

Model screening sensitivity is well-balanced across sex groups. Both Decision Tree and Random Forest models maintain ~65–68% recall across both women and men, consistently outperforming the clinical baseline rule (52.8% for women and 57.5% for men). The evaluated non-invasive ML models demonstrate equitable detection capability without exhibiting significant sex-based performance disparity on held-out test data.

Model sensitivity varies substantially across age cohorts. The recall advantage of machine learning over the clinical rule increases markedly with age:
1. In the 60–69 age bracket, ML model recall reaches 80.4%–82.8%, compared to 55.1% for the clinical baseline rule.
2. In the 50–59 age bracket, ML model recall reaches 62.4%–64.7%, compared to 54.5% for the clinical baseline rule.
3. In younger cohorts (30–39 and 40–49), ML recall drops to 55.7%–58.8%, performing closely to the clinical baseline rule.

Non-invasive clinical variables (bmi, pulse_pressure, age_years) provide strong predictive value in older demographics where cardiovascular disease prevalence is higher (51.9% at ages 50–59 and 67.0% at ages 60–69). However, lower sensitivity in adults under 50 suggests that non-invasive measurements alone may be insufficient for primary screening in younger populations, where incorporating invasive lab markers (cholesterol, gluc) may be clinically required. All subgroup evaluation metrics are computed strictly on the held-out test set (13722) to ensure valid generalization.
