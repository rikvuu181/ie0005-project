```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
sb.set()

df = pd.read_csv("../data/cardio_train.csv", sep=";")

print(df.head())
print(df.shape)
print(df.info())
print(df["cardio"].value_counts())

f, axes = plt.subplots(2, 3, figsize=(24, 12))

# Age vs cardio
df['age_years']=df['age']/365
sb.histplot(
    data=df, x="age_years", hue="cardio",
    bins=30, ax=axes[0, 0]
)

# Sex vs cardio
sb.countplot(
    data=df, x="gender", hue="cardio",
    ax=axes[0, 1]
)

# Weight vs cardio
sb.boxplot(
    data=df, x="cardio", y="weight",
    ax=axes[0, 2]
)

# Systolic blood pressure vs cardio
sb.boxplot(
    data=df, x="cardio", y="ap_hi",
    ax=axes[1, 0]
)

# Diastolic blood pressure vs cardio
sb.boxplot(
    data=df, x="cardio", y="ap_lo",
    ax=axes[1, 1]
)

# Smoking vs cardio
sb.countplot(
    data=df, x="smoke", hue="cardio",
    ax=axes[1, 2]
)

plt.show()
```
